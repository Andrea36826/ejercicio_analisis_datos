```python
import pandas as pd
```

```python
list_columns = ["Producto","Precio","Cantidad"]
list_data = [
    ["Manzanas" , 3000 , 10],
    ["Peras", 2500 , 5],
    ["Bananas", 2000 , 8],
    ["Mandarina", 3000, 12],
    ["Naranjas", 2000 , 20],
    ["Fresas", 2500 , 7]
]
df_data = pd.DataFrame(data=list_data , columns=list_columns)
display(df_data)
```

Creacion y modificacion de columnas 

```python
df_data.rename(columns={"Producto":"Productos"}, inplace=True)
display(df_data)
```

Eliminacion de una columna 

```python
df_data = df_data.drop(columns=["Cantidad"])
display(df_data)
```

Orden de datos

```python
df_data = df_data.sort_values("Precio", ascending=True)
display(df_data)
```

Manejo de valores faltantes

```python
df_data.loc[len(df_data)] = ["Mangos", None]
display(df_data)
```

```python
print(df_data.isnull().sum())
df_data = df_data.dropna(subset=["Precio"])
display(df_data)
```

```python
df_data.loc[len(df_data)] = ["Bananas", 0]
display(df_data)
```

```python
df_data["Precio"].fillna(0)
display(df_data)
```

Eliminar duplicados 

```python
df_data.loc[len(df_data)] = ["Bananas", 2000]
display(df_data)
```

```python
df_data.drop_duplicates(inplace=True)
display(df_data)
```
