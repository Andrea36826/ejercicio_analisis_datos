Trabajo en clase, importar archivo CSV

```python
df_personas = pd.read_csv(r"C:\Users\USUARIO\Downloads\Ejercicio de clase - Hoja 1.csv",header=1)
display(df_personas)
```

Eliminar columnas vacias 

```python
df_personas.drop(columns=["Unnamed: 1"] , inplace=True)
display(df_personas)
```

Quitar la tilde y el nombre de la tabla sin espacio 

```python
df_personas.rename(columns={'Género' : 'Genero','Nivel educativo':'Nivel_educativo'}, inplace=True)
display(df_personas)
```

elimnar nulo 

```python
df_personas = df_personas.dropna(subset=["Nombre","Apellido"])
display(df_personas)
```

```python
df_personas.info()
```

todos los string volverlo a none

```python
df_personas.loc[9, "Edad"] = None
df_personas.loc[18, "Edad"] = None
df_personas.loc[19, "Edad"] = None
df_personas.loc[25, "Edad"] = None
df_personas.loc[36, "Edad"] = None
display(df_personas)
```

```python
df_personas["Edad"] = df_personas["Edad"].astype("Float64")
df_personas["Edad"] = df_personas["Edad"].astype("Int64")
display(df_personas)
```

Promedio de la edad

```python
promedio_edad = round(df_personas["Edad"].mean(), 2)
print(promedio_edad)
```

Reemplazo de valores

```python
df_personas["Genero"].replace({"M":"Masculino", "F":"Femenino", "m":"Masculino", "f":"Femenino", "Indefinido": None}, inplace=True)
display(df_personas)
```

contar

```python
df_femeninos = df_personas[df_personas["Genero"] == "Femenino"]
display(df_femeninos)
cantidad_femenino = df_femeninos["Genero"].count()
print("La cantidad de mujeres es:", cantidad_femenino)
```

```python
df_masculinos = df_personas[df_personas["Genero"] == "Masculino"]
display(df_masculinos)
cantidad_masculino = df_masculinos["Genero"].count()
print("La cantidad de hombres es:", cantidad_masculino)
```

```python
data = df_personas.groupby("Genero")["Nombre"].count()
print(data)
```

Poner mayusculas 

```python
df_personas["Nombre"] = df_personas["Nombre"].str.capitalize()
df_personas["Apellido"] = df_personas["Apellido"].str.capitalize()
df_personas
```

```python
df_personas["Nivel_educativo"].replace({"universitario":"Universidad", "bachiller":"Bachillerato", "sec":"Bachillerato", "secundaria":"Bachillerato", "uni": "Universidad"} , inplace=True)
display(df_personas)
```

```python
print(df_personas["Nivel_educativo"].str.contains("Univer"))
```

```python
df_personas.loc[df_personas["Nivel_educativo"].str.contains("Uni", na=False), "Nivel_educativo"]= "Universidad"
df_personas.loc[df_personas["Nivel_educativo"].str.contains("Bach", na=False), "Nivel_educativo"]= "Bachillerato"
df_personas.loc[df_personas["Nivel_educativo"].str.contains("Sec", na=False), "Nivel_educativo"] = "Secundaria"
display(df_personas)
```

```python
df_personas.to_excel("data_clean.xlsx",index=False)
```
