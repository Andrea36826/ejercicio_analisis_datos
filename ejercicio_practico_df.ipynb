```python
import pandas as pd
```

Crear DataFrame 

```python
list_columns = ["Producto","Precio","Cantidad"]
list_data = [
    ["Manzanas" , 3000 , 10],
    ["Peras", 2500 , 5],
    ["Bananas", 2000 , 8],
    ["Mandarina", 3000, 12],
    ["Naranjas", 2000 , 20],
    ["Fresas", 2500 , 7]
]
df_data = pd.DataFrame(data=list_data , columns=list_columns)
display(df_data)
```

Agregar columna valor total 

```python

df_data["valor total"] = df_data["Precio"]* df_data["Cantidad"]
display(df_data)
```

Filtrar las ventas mayores a 20.000

```python
df_filtrar = df_data[df_data["valor total"] > 20000]
display(df_filtrar)
```

Determinar el producto con mayor ventas 

```python
pos_max = df_data["valor total"].idxmax()
print(df_data.loc[pos_max,"Producto"])
```

Agregar ua columna descuento 10 %

```python
df_data["Descuento"] = (df_data["valor total"] * 0.1).astype(int)
display(df_data)
```

Obtener ventas totales 

```python
total = df_data["valor total"].sum()
print(total)
```

Obetener promedio

```python
promedio_ventas = round(df_data["valor total"].mean(), 2)
print(promedio_ventas)
```

Obtener data frame de mayor a menor 

```python
df_ordenado = df_data.sort_values("valor total" , ascending=False)
display(df_ordenado)
```
